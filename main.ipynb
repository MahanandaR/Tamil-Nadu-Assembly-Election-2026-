# Final Constituency Summary Table

```python
# Select winner columns

winners_2021 = winners_2021[[
    'ac_number',
    'constituency',
    'candidate',
    'party',
    'votes',
    'turnout',
    'region',
    'reserved'
]]

winners_2026 = winners_2026[[
    'ac_number',
    'constituency',
    'candidate',
    'party',
    'votes',
    'turnout',
    'region',
    'reserved'
]]
```

```python
# Rename winner columns

winners_2021.columns = [
    'ac_number',
    'constituency',
    'winner_candidate',
    'winner_party',
    'winner_votes',
    'turnout',
    'region',
    'reserved'
]

winners_2026.columns = [
    'ac_number',
    'constituency',
    'winner_candidate',
    'winner_party',
    'winner_votes',
    'turnout',
    'region',
    'reserved'
]
```

```python
# Select runner-up columns

runners_2021 = runners_2021[[
    'ac_number',
    'candidate',
    'party',
    'votes'
]]

runners_2026 = runners_2026[[
    'ac_number',
    'candidate',
    'party',
    'votes'
]]
```

```python
# Rename runner-up columns

runners_2021.columns = [
    'ac_number',
    'runner_candidate',
    'runner_party',
    'runner_votes'
]

runners_2026.columns = [
    'ac_number',
    'runner_candidate',
    'runner_party',
    'runner_votes'
]
```

```python
# Merge winners and runner-ups

summary_2021 = winners_2021.merge(
    runners_2021,
    on='ac_number'
)

summary_2026 = winners_2026.merge(
    runners_2026,
    on='ac_number'
)
```

```python
# Calculate victory margin

summary_2021['margin'] = (
    summary_2021['winner_votes']
    - summary_2021['runner_votes']
)

summary_2026['margin'] = (
    summary_2026['winner_votes']
    - summary_2026['runner_votes']
)
```

```python
summary_2026.head()
```

###### Flip analysis helps identify constituencies
###### where the winning party changed between
###### 2021 and 2026 elections.
###### This helps analyze political shifts,
###### regional swings, and changing voter preferences.

```python
# Select required columns

flip_2021 = summary_2021[[
    'ac_number',
    'constituency',
    'winner_party'
]]

flip_2026 = summary_2026[[
    'ac_number',
    'winner_party'
]]
```

```python
# Rename columns

flip_2021.columns = [
    'ac_number',
    'constituency',
    'winner_2021'
]

flip_2026.columns = [
    'ac_number',
    'winner_2026'
]
```

```python
# Merge both election years

flip_df = flip_2021.merge(
    flip_2026,
    on='ac_number'
)
```

```python
# Create flip flag

flip_df['flipped'] = (
    flip_df['winner_2021'] != flip_df['winner_2026']
)
```

```python
flip_df.head()
```

```python
flip_df['flipped'].value_counts()  # Count total flipped seats.
```

```python
# Count seats won by each party in 2021

party_2021 = (
    summary_2021['winner_party']
    .value_counts()
    .reset_index()
)

party_2021.columns = [
    'party',
    'seats_2021'
]
```

```python
# Count seats won by each party in 2026

party_2026 = (
    summary_2026['winner_party']
    .value_counts()
    .reset_index()
)

party_2026.columns = [
    'party',
    'seats_2026'
]
```

```python
# Merge both election results

party_comparison = party_2021.merge(
    party_2026,
    on='party',
    how='outer'
).fillna(0)
```

```python
# Convert seat counts to integer

party_comparison['seats_2021'] = (
    party_comparison['seats_2021']
    .astype(int)
)

party_comparison['seats_2026'] = (
    party_comparison['seats_2026']
    .astype(int)
)
```

```python
# Calculate seat gain/loss

party_comparison['seat_change'] = (
    party_comparison['seats_2026']
    - party_comparison['seats_2021']
)
```

```python
# Sort by 2026 performance

party_comparison = (
    party_comparison
    .sort_values(
        by='seats_2026',
        ascending=False
    )
)

party_comparison
```

```python
# Region-wise seat count for 2026

region_party_2026 = (
    summary_2026
    .groupby(['region', 'winner_party'])
    .size()
    .reset_index(name='seats')
)

region_party_2026.head()
```

```python
# Sort for readability

region_party_2026 = (
    region_party_2026
    .sort_values(
        ['region', 'seats'],
        ascending=[True, False]
    )
)

region_party_2026
```

```python
# Find top party in each region

regional_winner = (
    region_party_2026
    .sort_values(
        ['region', 'seats'],
        ascending=[True, False]
    )
    .drop_duplicates('region')
)

regional_winner
```

```python
# Top 10 closest contests in 2026

closest_seats = (
    summary_2026
    .sort_values('margin')
    .head(10)
)

closest_seats[[
    'constituency',
    'winner_party',
    'runner_party',
    'margin',
    'region'
]]
```

```python
closest_seats = closest_seats.reset_index(drop=True)

closest_seats[[
    'constituency',
    'winner_party',
    'runner_party',
    'margin',
    'region'
]]
```

```python
# Save processed datasets

summary_2021.to_csv(
    "data/summary_2021.csv",
    index=False
)

summary_2026.to_csv(
    "data/summary_2026.csv",
    index=False
)

flip_df.to_csv(
    "data/flip_analysis.csv",
    index=False
)

party_comparison.to_csv(
    "data/party_comparison.csv",
    index=False
)

region_party_2026.to_csv(
    "data/region_party_2026.csv",
    index=False
)

closest_seats.to_csv(
    "data/closest_contests.csv",
    index=False
)
```
